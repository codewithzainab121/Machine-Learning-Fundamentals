## Solved Examples

### Example 1: Replace a Single Value with `NaN`

```python
import pandas as pd
import numpy as np
```

```python
# Create a sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Dave'],
    'Age': [25, -99999, 30, -99999],
    'Salary': [50000, -99999, 60000, 70000]
}
df = pd.DataFrame(data)
```

```python
# Replace -99999 with NaN

df_replaced = df.replace(-99999, np.nan)
print("After replacing -99999 with NaN:")
print(df_replaced)
```

# Example 2: Replace Multiple Values Using a Dictionary
### Replace specific values in specific columns

```python
replace_dict = {
    'Age': {-99999: np.nan},
    'Salary': {-99999: np.nan, 70000: 75000}
}
df_replaced_dict = df.replace(replace_dict)
print("\nAfter replacing values using a dictionary:")
print(df_replaced_dict)
```

# Example 3: Using `regex` to Replace Values
### Create a sample DataFrame with strings

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Dave'],
    'Email': ['alice@example.com', 'bob@test.com', 'charlie@demo.com', 'dave@sample.com']
}
df_strings = pd.DataFrame(data)
```

```python
# Replace email domains using regex
df_regex = df_strings.replace({'Email': r'@.*\.'}, '@newdomain.', regex=True)
print("\nAfter replacing email domains using regex:")
print(df_regex)
```

### Problem 1: Replace Missing Values

```python
import numpy as np

data = {
    'Name' : ['Zainab' , 'Nida' , 'Noor' , 'minhail'],
    'Age'  : [21, -8 , 17 , 22],
    'Score' : [100, 93, 101 , -55]
}
df = pd.DataFrame(data)
df.replace(to_replace = [-8, 101, -55], value = 0 , inplace=True)
print(df)
```

### Problem 2: Replace Values in Specific Columns
Using the same DataFrame from Problem 1, replace `-99999` with `NaN` only in the `Age` column and replace `-1` with `0` only in the `Salary` column.

```python
# Create a sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Dave'],
    'Age': [25, -99999, 30, -99999],
    'Salary': [50000, -1, 60000, 70000]
}
df = pd.DataFrame(data)
df.replace({
    'Age' : {
        -99999 : np.nan
},
    'Salary' : {
        -1 : 0
    }
}, inplace = True)
df
```

### Problem 4: Replace Using `regex`
Create a DataFrame with the following data:

Replace all email domains with `@newdomain.com` using `regex`.

```python
data = {
    'Message': [
        "Contact us at info@example.com",
        "Support: support@test.com",
        "Feedback: feedback@demo.com"
    ]
}
df = pd.DataFrame(data)
df['Message'] = df['Message'].str.replace(r'@\w+\.\w+', '@newdomain.com', regex=True)
df
```

### Problem 6: Replace with Conditions
Create a DataFrame with the following data:

Replace all scores less than `50` with `Fail` and scores greater than or equal to `50` with `Pass`.

```python
df = pd.DataFrame({
    'Name' : ['Alice','Bob','Charlie', 'Dave'],
    'Score' : [85 , 45 , 90 , 50]
})

df['Score'] = df['Score'].apply(lambda x: "Pass" if x >= 50 else "Fail")

print(df)
```
