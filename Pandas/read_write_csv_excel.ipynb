### Read CSV</color>

```python
import pandas as pd
```

```python
df = pd.read_excel("stocks_weather.xlsx")
df
```

```python
df1 = df.drop(columns=['Unnamed: 0'])
display(df1)
```

```python
# This cell now uses the 'df1' DataFrame, which was created by removing the 'Unnamed: 0' column from the original 'df'
print("Continuing with the updated df1 DataFrame:")
display(df1)
```

```python
# To remove the first row of df1:
df1_skipped_row = df1.iloc[1:]

print("df1 after skipping the first row:")
display(df1_skipped_row)
```

This creates a new DataFrame `df1_skipped_row` that contains all rows from `df1` except for the very first one.
