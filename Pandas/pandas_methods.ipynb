## 1. Convert String Column into Date Type



```python

import pandas as pd

# Sample DataFrame
data = {
    'date': ['2023-10-01', '2023-10-02', '2023-10-03'],
    'value': [10, 20, 30]
}
df = pd.DataFrame(data)
print(type(df.date[0]))

# Convert 'date' column to datetime
df['date'] = pd.to_datetime(df['date'])
print(df)
print(type(df.date[0]))
```

# 2. Use Date as an Index of DataFrame Using set_index()
Example Problem:
Set the date column as the index of the DataFrame.

```python
# Sample DataFrame
data = {
    'date': ['2023-10-01', '2023-10-02', '2023-10-03'],
    'value': [10, 20, 30]
}
df = pd.DataFrame(data)
df['date'] = pd.to_datetime(df['date'])

# Set 'date' column as index
df.set_index('date', inplace=True)
print(df)
```

# 3. Use fillna() Method in DataFrame
Example Problem:
Replace missing values in the value column with 0.

```python
import numpy as np

# Sample DataFrame with missing values
data = {
    'date': pd.to_datetime(['2023-10-01', '2023-10-02', '2023-10-03']),
    'value': [10, np.nan, 30]
}
df = pd.DataFrame(data)

# Replace NaN with 0
df['value'].fillna(0, inplace=True)
print(df)
```

# Practice Problem:
Replace missing values in the value column with the mean of the column.

```python
data = {
    'date': pd.to_datetime(['2023-10-01', '2023-10-02', '2023-10-03']),
    'value': [10, np.nan, 30]
}
df = pd.DataFrame(data)
df['value'].fillna(df['value'].mean(), inplace=True)
print(df)
```

# 4. Use fillna(method="ffill") Method in DataFrame
Example Problem:
Forward fill missing values in the value column.

```python
data = {
    'date': pd.to_datetime(['2023-10-01', '2023-10-02', '2023-10-03']),
    'value': [10, np.nan, 30]
}
df = pd.DataFrame(data)
df['value'].fillna(method='ffill', inplace=True)
print(df)
```

# 5. Use fillna(method="bfill") Method in DataFrame
Example Problem:
Backward fill missing values in the value column.

```python
data = {
    'date': pd.to_datetime(['2023-10-01', '2023-10-02', '2023-10-03']),
    'value': [10, np.nan, 30]
}
df = pd.DataFrame(data)
df['value'].fillna(method='bfill', inplace=True)
print(df)
```

# Practice Problem:

# 6. "axis" Parameter in fillna() Method
Example Problem:
Replace missing values column-wise (axis=1)

```python
# Column-wise fillna example on a small demo DataFrame
df_axis = pd.DataFrame({'A': [1, None, 3],
                        'B': [None, 5, None]
                        })
df_axis.fillna(method = 'ffill', axis=1, inplace=True)
print(df_axis)
```

# Practice Problem:
Replace missing values row-wise (axis=0) in the following DataFrame.

```python
df_axis = pd.DataFrame({
    'A': [1, None, 3],
    'B': [None, 5, None]
    })
df_axis.fillna(method = 'ffill' , axis=0, inplace=True)
print(df_axis)
```

# 7. "limit" Parameter in fillna() Method
Example Problem:
Replace only the first missing value in the value column.

```python
data = {
    'date': pd.to_datetime(['2023-10-01', '2023-10-02', '2023-10-03']),
    'value': [10, np.nan, 30]
}
df = pd.DataFrame(data)
df['value'].fillna(15, limit=1, inplace=True)
print(df)
```

# 8. interpolate() to Do Interpolation in DataFrame
Example Problem:
Interpolate missing values in the value column.

```python
data = {
    'date': pd.to_datetime(['2023-10-01', '2023-10-02', '2023-10-03', '2023-10-04', '2023-10-05']),
    'value': [10, np.nan, 30, np.nan, np.nan]
}
df = pd.DataFrame(data)
df['value'].interpolate(inplace=True)
print(df)
```

# Practice Problem:
Interpolate missing values in the value column of the following DataFrame using time-based interpolation.

```python
data = {
    'date': pd.to_datetime(['2023-10-01', '2023-10-03', '2023-10-06']),
    'value': [10, np.nan, 40]
}
df = pd.DataFrame(data)

# Set the date column as the index for time-based interpolation
df = df.set_index('date')

# Interpolate missing values using time-based interpolation
df['value'].interpolate(method='time', inplace=True)
print(df)
```

# 10. dropna() Method to Drop Rows with Missing Values
Example Problem:
Drop all rows with missing values.

```python
data = {
    'date': pd.to_datetime(['2023-10-01', '2023-10-03', '2023-10-06']),
    'value': [10, np.nan, 40]
}
df = pd.DataFrame(data)
df.dropna(inplace=True)
print(df)
```

# Practice Problem:
Drop all rows with missing values in the following DataFrame.

```python
data = {
    'date': pd.to_datetime(['2023-10-01', '', '2023-10-06']),
    'value': [10, np.nan, 40],
    'temperature' : [20, np.nan, np.nan]
}
df = pd.DataFrame(data)
df.dropna(how='all', inplace=True)
df
```

# 11. "how" Parameter in dropna() Method
Example Problem:
Drop rows where all values are missing.

```python
df.dropna(how='all', inplace=True)
print(df)
```

# Practice Problem:
Drop rows where any value is missing in the following DataFrame.

```python
data = {
    'date': pd.to_datetime(['2023-10-01', '', '2023-10-06']),
    'value': [10, np.nan, 40],
    'temperature' : [20, np.nan, np.nan]
}
df = pd.DataFrame(data)
df.dropna(how='any', inplace=True)
df
```

# 12. "thresh" Parameter in dropna() Method
Example Problem:
Drop rows with fewer than 2 non-missing values.

```python
df.dropna(thresh=2, inplace=True)
print(df)
```
